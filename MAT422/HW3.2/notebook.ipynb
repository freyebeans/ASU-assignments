```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)
```

## 3.2.1 Limits and Continuity
A limit $L = \lim_{x\to a}f(x)$ is what describes a function $f$'s behavior as $x$ approaches $a$. A function $f$ is said to be continuous if $\lim_{x\to a}f(x)=f(a)$. A continuous function does not make any abrupt changes (called discontinuties) as $x$ varies.

```python
h = np.array([1, 0.5, 0.1, 0.01, 0.001])
values = (np.sin(h) ** 2) / h
print("h values:", h)
print("sin^2(h)/h:", values)
print("These values approach 0 as h approaches 0.")

x = np.linspace(-3, 3, 400)
y = np.where(x < 0, np.sin(x), np.cos(x))
plt.plot(x, y)
plt.axvline(0, color="black", linewidth=0.8)
plt.title("Example of a jump discontinuity at x = 0")
plt.show()
```

As you can see, $\lim_{h\to 0}\sin^2(h) = 0$, which is to say that as $h$ approaches 0, $\sin^2(h)$ also approaches 0. additionally you can see that the above function is not continuous as at x = 0 the function jumps from 0 to 1 abruptly.
## 3.2.2 Derivatives
A derivative measures the way the output of a function $f$ changes with respect to the input $x$. The derivative of $f$ at $x$ is defined as $f'(x)=\lim_{h\to 0}\frac{f(x+h)-f(x)}{h}$. $f'(x)$ can also be expressed as $f'(x)=\lim_{h\to 0}\frac{f(x)-f(x-h)}{h}$ and $f'(x)=\lim_{h\to 0}\frac{f(x+h)-f(x-h)}{2h}$.

```python
def f(x):
    return np.sin(x)

point = np.pi/6
steps = np.array([1e-1, 1e-2, 1e-3, 1e-4])
numeric_derivatives = (f(point + steps) - f(point - steps)) / (2 * steps)
exact_derivative = np.cos(point)

print("numeric derivative estimates:", numeric_derivatives)
print("exact derivative:", exact_derivative)
print("errors:", np.abs(numeric_derivatives - exact_derivative))
```

Many functions can be differentiated (have their derivative taken) analytically. One such function is the sine of x, whose derivative is the cosine of x. As you can see above, numerically evaluating $f'(x)=\lim_{h\to 0}\frac{f(x+h)-f(x-h)}{2h}$ where $f(x) = \sin(x)$ and $x=\pi/6$ yields values that approach the cosine of x as h approaches 0.
## 3.2.3 Taylor's Theorm
Taylor's theorm gives a polynomial approximation of a differentiable function $f$ around a particular point. For a function $f(x)$ that is $n$-times differentiable on the interval from $a$ to $x$, we can find a Taylor polynomial $P_n(x) = \sum^{n}_{k=0}\frac{(x-a)^kf^{(k)}(a)}{k!}$ up to degree $n$ approximating $f(x)$ around $a$. The error of this approximation $R_n = f(x) - P_n(x)$ approaches 0 as $x$ approaches $a$.

```python
x = np.linspace(-2, 2, 300)
taylor_sin = x - x**3 / 6

plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, taylor_sin, "--", label="third-degree Taylor polynomial at 0")
plt.xlabel("x")
plt.ylabel("value")
plt.title("Taylor approximation")
plt.legend()
plt.show()

print("Max error on [-1, 1] =", np.max(np.abs(np.sin(x[np.abs(x) <= 1]) - taylor_sin[np.abs(x) <= 1])))
```

As you can see, the maximum error on the interval [-1,1] is quite small, meaning that for this interval, the Taylor polynomial closely approximates the sine of $x$.
